```python
# ================================================================
#   ANÁLISE DE PROJEÇÃO DE TRAJETÓRIAS DE DINÂMICA MOLECULAR
#   (Versão descritiva - Figura tipo S17, sem o sistema AG73/Sind)
#
#   Objetivo:
#   Este script projeta trajetórias de dinâmica molecular (MD)
#   nos dois primeiros modos normais ou componentes principais
#   (mode 1 e mode 2), permitindo visualizar o espaço conformacional
#   explorado durante as simulações.
#
#   Sistemas considerados:
#       - Apo (integrina sem ligante)
#       - AG73/Sind (integrina complexada com ligante AG73/Sind)
#
#   Cada réplica da simulação é representada por um conjunto de pontos
#   em cores distintas (tons de azul e laranja), mostrando variação
#   conformacional entre simulações independentes.
#
#   Desenvolvido para uso direto no Google Colab.
# ================================================================

# ------------------------------------------------
# Importação das bibliotecas necessárias
# ------------------------------------------------
import pandas as pd                  # Manipulação de tabelas e dados numéricos
import matplotlib.pyplot as plt      # Geração de gráficos
import seaborn as sns                # Estética e estilo dos gráficos
from google.colab import files       # Upload de arquivos diretamente no Colab
```

```python
# ------------------------------------------------
# Etapa 1: Upload dos arquivos de projeção
# ------------------------------------------------
print("📂 Faça upload dos arquivos CSV contendo as colunas: mode1, mode2")
uploaded = files.upload()

# Cada arquivo deve conter duas colunas numéricas:
# mode1,mode2
#  1.23,-0.45
#  0.87,-1.12
#  ...
```

```python
# ------------------------------------------------
# Etapa 2: Leitura dos arquivos em dataframes do pandas
# ------------------------------------------------
data = {}
for name in uploaded.keys():
    df = pd.read_csv(name)
    data[name] = df

# ------------------------------------------------
# Etapa 3: Definição das cores para os sistemas estudados
# ------------------------------------------------
# Tons de azul → sistema Apo (sem ligante)
# Tons de laranja → sistema AG73/Synd (complexado)
colors = {
    'apo': ['#a6cee3', '#1f78b4', '#08306b'],       # tons de azul
    'C16S1': ['#fdbf6f', '#ff7f00', '#b35900'],     # tons de laranja
}

# ------------------------------------------------
# Etapa 4: Função auxiliar para identificar o sistema
# ------------------------------------------------
# Essa função procura no nome do arquivo termos como “apo” ou “AG73/Synd”
# e usa isso para definir a cor correspondente no gráfico.
def identify_system(filename):
    for key in colors.keys():
        if key.lower() in filename.lower():
            return key
    return 'unknown'

# ------------------------------------------------
# Etapa 5: Preparação da figura e dos subplots
# ------------------------------------------------
# O número de painéis (subplots) será igual ao número de arquivos enviados.
fig, axes = plt.subplots(1, len(data), figsize=(6 * len(data), 6))
if len(data) == 1:
    axes = [axes]  # garante que axes seja sempre uma lista, mesmo com 1 arquivo

```

```python
# ------------------------------------------------
# Etapa 6: Plotagem dos dados de cada réplica
# ------------------------------------------------
for i, (filename, df) in enumerate(data.items()):
    # Identifica qual sistema o arquivo representa
    system = identify_system(filename)

    # Seleciona a paleta de cores apropriada
    pal = colors.get(system, ['#aaaaaa'])
    color = pal[i % len(pal)]  # alterna tons entre réplicas

    # Seleciona o subplot correspondente
    ax = axes[i]

    # Gera o gráfico de dispersão (mode1 vs mode2)
    ax.scatter(df['mode1'], df['mode2'], s=10, color=color, alpha=0.6)

    # Linhas pontilhadas cruzando a origem (0,0)
    ax.axhline(0, color='k', ls='--', lw=1)
    ax.axvline(0, color='k', ls='--', lw=1)

    # Define limites e rótulos dos eixos
    ax.set_xlim(-8, 8)
    ax.set_ylim(-8, 8)
    ax.set_xlabel('mode 1', fontsize=12)
    ax.set_ylabel('mode 2', fontsize=12)

    # Define título do painel
    ax.set_title(f"{system.upper()} - réplica {i+1}",
                 fontsize=13, fontweight='bold')
```

```python
# ------------------------------------------------
# Etapa 7: Ajustes finais e exibição da figura
# ------------------------------------------------
plt.tight_layout()
plt.show()
```
